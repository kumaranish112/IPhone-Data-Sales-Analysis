```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python

                              # TOP 10 HIGHEST-RATED IPHONES ON FLIPCART IN INDIA
```

```python
data.head()
```

```python
highest_rated = data.sort_values(by = ["Star Rating"], ascending = False)
highest_rated = highest_rated.head(10)
```

```python
print(highest_rated["Product Name"])
```

```python
print(highest_rated)
```

```python

                              # HOW MANY RATINGS DO THE HIGHEST-RATED IPHONES ON FLIPKART HAVE?
```

```python
data.head()
```

```python
iphones = highest_rated["Product Name"].value_counts()
label = iphones.index
counts = highest_rated["Number Of Ratings"]
figure = px.bar(highest_rated, x=label, y = counts, title = "number of Ratings of Highest Rated iPhones")
figure.show()
```

```python

                                # WHICH IPHONE HAS THE HIGHEST NUMBER OF REVIEWS ON FLIPKART?
```

```python
data.head()
```

```python
iphones = highest_rated["Product Name"].value_counts()
label = iphones.index
counts = highest_rated["Number Of Reviews"]
figure = px.bar(highest_rated, x=label, y = counts, title = "number of reviews of Highest Rated iPhones")
figure.show()
```

```python

                   # WHAT IS THE RELATIONSHIP BETWEEN THE SALE PRICE OF IPHONES AND THE NUMBER OF RATINGS ON FLIPKART?
```

```python
data.head()
```

```python
figure = px.scatter(data_frame = data, x = "Number Of Ratings", y = "Sale Price", size = "Discount Percentage", trendline = "ols", title="relationship between price and number of rating")
figure.show()
```

```python
# their is a negative relationship between the sale price of iPhones and the number of ratings. It means iPhones with lower sale prices are sold more in india.
```

```python

            # WHAT IS THE RELATIONSHIP BETWEEN THE DISCOUNT PRECENTAGE AND THE NUMBER OF RATINGS OF IPHONES ON FLIPKART?
```

```python
figure = px.scatter(data_frame = data, x = "Number Of Ratings", y = "Discount Percentage", size = "Sale Price", trendline = "ols", title="relationship between Discount and number of rating")
figure.show()
```

```python

           # THE LEAST EXPENSIVE AND MOST EXPENSIVE IPHONES IN THE INDIAN MARKET, ALONG WITH ALL THEIR SPECIFICATIONS?
```

```python
data.head()
```

```python
most_expensive = data.loc[data['Sale Price'].idxmax()]
least_expensive = data.loc[data['Sale Price'].idxmax()]

# Display the results
print("Most Expensive Product:")
print(most_expensive)

print("\nLeast Expensive Product:")
print(least_expensive)
```
